```python

import re

variables = {'x', 'y', 'z', 'u', 'v', 'w'}

def parse(expr):
    tokens = re.findall(r'[A-Za-z_]\w*|[(),]', expr)
    pos = [0]

    def read():
        name = tokens[pos[0]]
        pos[0] += 1

        if pos[0] < len(tokens) and tokens[pos[0]] == '(':
            pos[0] += 1
            args = []

            while True:
                args.append(read())
                if tokens[pos[0]] == ')':
                    pos[0] += 1
                    break
                pos[0] += 1

            return (name, args)

        return name

    return read()


def is_variable(term):
    return isinstance(term, str) and term in variables


def apply(term, subs):
    if isinstance(term, str):
        while term in subs:
            term = subs[term]
        return term

    return (term[0], [apply(a, subs) for a in term[1]])


def occurs(var, term, subs):
    term = apply(term, subs)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, a, subs) for a in term[1])

    return False


def unify(a, b, subs):
    a = apply(a, subs)
    b = apply(b, subs)

    if a == b:
        return True

    if is_variable(a):
        if occurs(a, b, subs):
            return False
        subs[a] = b
        return True

    if is_variable(b):
        if occurs(b, a, subs):
            return False
        subs[b] = a
        return True

    if isinstance(a, tuple) and isinstance(b, tuple):
        if a[0] != b[0] or len(a[1]) != len(b[1]):
            return False

        for x, y in zip(a[1], b[1]):
            if not unify(x, y, subs):
                return False

        return True

    return False


def display(term):
    if isinstance(term, str):
        return term

    return term[0] + "(" + ", ".join(display(x) for x in term[1]) + ")"


# Main program
e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

a = parse(e1)
b = parse(e2)

subs = {}

if unify(a, b, subs):
    print("\nUnification Successful")
    print("Substitutions:")

    for key, value in subs.items():
        print(key, "=", display(apply(value, subs)))

    print("Unified Expression:", display(apply(a, subs)))
else:
    print("\nUnification Failed")
```
