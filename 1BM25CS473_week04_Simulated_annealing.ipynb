```python

import random
import math

def calculate_cost(state):
    h = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if abs(state[i] - state[j]) == abs(i - j):
                h += 1

    return h


def simulated_annealing(state):
    T = 100.0
    cooling_rate = 0.99
    step = 1

    while T > 0.1:
        current_cost = calculate_cost(state)

        print("Step:", step, "State:", state, "h =", current_cost)

        if current_cost == 0:
            print("Solution Found!")
            return state

        i, j = random.sample(range(8), 2)

        neighbor = state[:]
        neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

        next_cost = calculate_cost(neighbor)
        delta_E = current_cost - next_cost

        if delta_E > 0:
            state = neighbor
        else:
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                state = neighbor

        T *= cooling_rate
        step += 1

    if calculate_cost(state) == 0:
        print("Solution Found!")
    else:
        print("Temperature limit reached. Restarting...")

        state = random.sample(range(8), 8)
        return simulated_annealing(state)

    return state


def display_board(state):
    for row in range(8):
        for col in range(8):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


state = list(map(int, input(
    "Enter 8 queen row positions (0-7): "
).split()))

if len(state) != 8 or sorted(state) != list(range(8)):
    print("Invalid input! Enter each number from 0 to 7 once.")
else:
    solution = simulated_annealing(state)

    print("\nFinal solution:", solution)
    print("Final cost h =", calculate_cost(solution))
    display_board(solution)
```
