```python

from itertools import product

# Initialize propositions
symbols = ['P', 'Q', 'R']

# Logical operations
def implies(a, b):
    return (not a) or b

# Evaluate Knowledge Base
def evaluate_kb(model):
    P = model['P']
    Q = model['Q']
    R = model['R']

    return (implies(Q, P) and
            implies(P, not Q) and
            (Q or R))

# Evaluate Query
def evaluate_query(query, model):
    P = model['P']
    Q = model['Q']
    R = model['R']

    if query == 'R':
        return R

    elif query == 'R -> P':
        return implies(R, P)

    elif query == 'Q -> R':
        return implies(Q, R)

    else:
        return False

# Truth Table Enumeration
def tt_entails(query):

    result = True
    models = []

    print("\nP\tQ\tR\tKB\tQuery")

    for values in product([False, True], repeat=3):

        model = dict(zip(symbols, values))

        kb = evaluate_kb(model)
        q = evaluate_query(query, model)

        print(
            model['P'], "\t",
            model['Q'], "\t",
            model['R'], "\t",
            kb, "\t",
            q
        )

        if kb:
            models.append(model)

            if not q:
                result = False

    print("\nModels of KB:")
    for model in models:
        print(model)

    if result:
        print("\nKB entails", query)
    else:
        print("\nKB does not entail", query)

    return result


# Main Program
print("TRUTH TABLE ENUMERATION")

queries = ['R', 'R -> P', 'Q -> R']

for query in queries:
    print("\n==============================")
    print("QUERY:", query)
    tt_entails(query)
```
