```python

from heapq import *

goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def h(s):
    d = 0
    for i in range(9):
        if s[i] != 0:
            j = goal.index(s[i])
            r1, c1 = divmod(i, 3)
            r2, c2 = divmod(j, 3)
            d += abs(r1 - r2) + abs(c1 - c2)
    return d

def astar(start):
    pq = [(h(start), 0, start, [])]
    visited = set()

    while pq:
        f, g, s, path = heappop(pq)

        if s == goal:
            return path + [s]

        if s in visited:
            continue
        visited.add(s)

        z = s.index(0)

        for d in [-3, 3, -1, 1]:
            n = z + d

            if n < 0 or n > 8:
                continue
            if d == -1 and z % 3 == 0:
                continue
            if d == 1 and z % 3 == 2:
                continue

            a = list(s)
            a[z], a[n] = a[n], a[z]
            a = tuple(a)

            if a not in visited:
                heappush(pq, (g + 1 + h(a), g + 1, a, path + [s]))

    return None

start = tuple(map(int, input("Enter 9 numbers (0 for blank): ").split()))
result = astar(start)

if result:
    for s in result:
        print(s[:3], s[3:6], s[6:])
else:
    print("No solution")
```
