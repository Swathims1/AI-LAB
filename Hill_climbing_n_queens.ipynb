```python
def calculate_cost(state):
    h = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                h += 1

    return h


def hill_climbing(state):
    n = len(state)

    while True:
        current_cost = calculate_cost(state)

        print("State:", state, "h =", current_cost)

        if current_cost == 0:
            print("Solution found!")
            return state

        best_state = state[:]
        best_cost = current_cost

        for i in range(n):
            for j in range(i + 1, n):
                neighbor = state[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                cost = calculate_cost(neighbor)

                if cost < best_cost:
                    best_cost = cost
                    best_state = neighbor

        if best_cost >= current_cost:
            print("Local minimum reached. Restarting...")
            state = [2, 0, 3, 1]
        else:
            state = best_state


def display_board(state):
    n = len(state)

    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


n = 4
state = list(map(int, input("Enter 4 queen row positions (0-3): ").split()))

if len(state) != n or sorted(state) != list(range(n)):
    print("Invalid input! Enter a permutation of 0 1 2 3.")
else:
    solution = hill_climbing(state)
    print("Final solution:", solution)
    print("Final cost h =", calculate_cost(solution))
    display_board(solution)
```
